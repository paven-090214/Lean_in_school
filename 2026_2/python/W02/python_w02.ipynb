```python
import pandas as pd
df1 = pd.DataFrame(
    {
        'name': ['안수환', '추교현', '마성혁', '이시현'],
        'database' : ['A+', 'A', 'B', 'C'],
        'java': [50, 60, 70, 80]
    }
)
print(df1)
```

```python
sum(df1['java']) / len(df1)
```

```python
# [] = 컬럼이 한 개인 Series 자료구조를 사용, column이 한 개일 경우
# df1['name','java'] -> 오류
# [[]] = DataFrame으로 변경, column이 한 개 이상
df1[['name','java', 'database']]
```

```python
result = df1 ['java'] >= 70
# ture, false의 값으로 결과가 나옴, column 이름이 존재하지 않음. 1차원 구조
print(result)
```
